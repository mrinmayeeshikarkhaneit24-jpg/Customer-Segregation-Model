```python
X = df[['Annual Income (k$)', 'Spending Score (1-100)']]
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
X_scaled[:5]
```

```python
inertia = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    range(2, 11),
    inertia,
    marker='o',
    color = 'red'
)

plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal K')

plt.show()
```

```python
kmeans = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(X_scaled)
df['Cluster'] = clusters
df.head()
```

```python
plt.figure(figsize=(10,7))

sns.scatterplot(
    data=df,
    x='Annual Income (k$)',
    y='Spending Score (1-100)',
    hue='Cluster',
    palette='Set1',
    s=100
)

plt.title('Customer Segmentation using K-Means')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')

plt.show()
```

```python
score = silhouette_score(X_scaled, clusters)

print("Silhouette Score:", score)
```

```python
silhouette_scores = []

for k in range(2, 11):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_scaled)

    score = silhouette_score(X_scaled, labels)

    silhouette_scores.append(score)
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    range(2, 11),
    silhouette_scores,
    marker='o',
    color = 'purple'
)

plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score for Different K Values')

plt.show()
```

```python
cluster_summary = df.groupby('Cluster')[
    ['Age', 'Annual Income (k$)', 'Spending Score (1-100)']
].mean()

cluster_summary
```

```python
plt.figure(figsize=(10,7))

sns.scatterplot(
    data=df,
    x='Annual Income (k$)',
    y='Spending Score (1-100)',
    hue='Cluster',
    palette='Set1',
    s=120
)

plt.title('Final Customer Segmentation')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend(title='Customer Segment')

plt.show()
```

```python
for k, score in zip(range(2, 11), silhouette_scores):
    print(f"K = {k}: Silhouette Score = {score:.4f}")

best_k = list(range(2, 11))[
    silhouette_scores.index(max(silhouette_scores))
]

print("\nBest K by Silhouette Score:", best_k)
print("Highest Silhouette Score:", round(max(silhouette_scores), 4))
```

```python
cluster_counts = df['Cluster'].value_counts().sort_index()

print("Number of customers in each cluster:")
print(cluster_counts)

cluster_counts.plot(
    kind='bar',
    figsize=(8, 5),
    color='teal',
    edgecolor='black'
)

plt.xlabel("Cluster")
plt.ylabel("Number of Customers")
plt.title("Customer Count in Each Cluster")
plt.xticks(rotation=0)
plt.show()
```

```python
cluster_names = {
    0: "Average Income - Average Spending",
    1: "High Income - High Spending",
    2: "Low Income - High Spending",
    3: "High Income - Low Spending",
    4: "Low Income - Low Spending"
}

df['Segment'] = df['Cluster'].map(cluster_names)

df[['CustomerID', 'Annual Income (k$)',
    'Spending Score (1-100)', 'Cluster', 'Segment']].head(10)
```

```python
# Save the customer segmentation results
df.to_csv("Customer_Segmentation_Results.csv", index=False)

print("Customer segmentation results saved successfully!")
```
